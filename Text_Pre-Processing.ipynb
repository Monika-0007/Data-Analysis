```python
#Importing NLP libraries
import nltk
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')
from nltk import ne_chunk
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
import textblob, string, re  #For text processing and regex operations
from nltk.corpus import stopwords  #For removing common words
from textblob import TextBlob, Word  #For sentiment analysis and word processing
from nltk.stem import PorterStemmer, WordNetLemmatizer  #For word normalization
from bs4 import BeautifulSoup  #For parsing HTML content

#Importing data visualization and manipulation libraries
import matplotlib.pyplot as plt  # For creating plots and visualizations
import pandas as pd  # For data manipulation and analysis
import numpy as np  # For numerical operations

#Suppress warning messages for clean output 
from warnings import filterwarnings
filterwarnings('ignore' )
```

```python
text= """
I can't wait for the Season 5 of Stranger Things🤩❤!
The global AI market is valued at over $390 billion and is projected to increase in value by over 5x in the next 5 years😲. 
'Friends, Romans, countrymen, lend me your ears: I come to bury Caesar, not to praise him.'
<html><body>Welcome to Text Pre-processing!</body></html>
"🐍Python" is a great programming language!!! ??
"""
```

```python
print(text)
```

```python
#Splits the text based on newline
sentences_text = text.split("\n")
#Converts the list of sentences to a pandas Series
text_to_series = pd.Series(sentences_text)
#Extracts all sentences except the first and last (which might be empty or headers)
text_vector = text_to_series[1:len(text_to_series)-1]
#Create DataFrame with the sentences, using "sentences" as the column name
tdf = pd.DataFrame(text_vector, columns = ["sentences"])
```

```python
#Converts all text to lowercase
tdf["lower_case"] = tdf["sentences"].apply(lambda x: " ".join(x.lower() for x in x.split()))
#Removes all punctuation marks from the text
tdf["punctuations_removed"] = tdf["lower_case"].apply(lambda x: "".join([letter for letter in x if letter not in string.punctuation]))
#Removes all numeric digits from the text using regex
tdf["numbers_removed"] = tdf["punctuations_removed"].str.replace("\\d", "", regex=True)
```

```python
!pip install emoji
import emoji
#Removes all emojis from the text
tdf["emojis_removed"] = tdf["numbers_removed"].apply(lambda text: "".join(c for c in text if c not in emoji.EMOJI_DATA))
```

```python
# Get a list of English stopwords (common words like 'the', 'a', 'an', etc.)
sw = stopwords.words("english")
tdf["stopwords_removed"] = tdf["emojis_removed"].apply(lambda x: " ".join(word for word in x.split() if word not in sw))
```

```python
#Tokenization - breaks down a piece of text into smaller units called tokens
tdf["tokenized"] = tdf["stopwords_removed"].apply(lambda x: TextBlob(x).words)
#Stemming - words are stemmed or diminished to their root form
tdf["stemmed"] = tdf["stopwords_removed"].apply(lambda x: " ".join([PorterStemmer().stem(word) for word in x.split()]))
tdf
```

```python
#Lemmatization - reduces the inflected words properly ensuring the root word belongs to the language.

tdf["lemmatized"] = tdf["stopwords_removed"].apply(lambda x: " ".join([WordNetLemmatizer().lemmatize(word) for word in x.split()]))
tdf.iloc[:, 5:]
```

```python
# PART OF SPEECH (POS) Tagging - assigns grammatical categories to each word in a text. 

from nltk.tokenize import word_tokenize
from nltk import pos_tag
 
# Performing POS tagging
def pos_tagging(text):
    word_token = word_tokenize(text)
    return pos_tag(word_token)
    
print("Original Text:", tdf["sentences"].iloc[1])
print("\nPOS Tagging Result:",pos_tagging(tdf["sentences"].iloc[1]))
```

```python
# CHUNKING - process of identifying parts of speech and short phrases in a sentence.

def chunking(text, grammar):
    word_tokens = word_tokenize(text)
    # Label words with part of speech
    word_pos = pos_tag(word_tokens)
    # create a chunk parser using grammar
    chunkParser = nltk.RegexpParser(grammar)
    # test it on the list of word tokens with tagged pos
    tree = chunkParser.parse(word_pos)
    
    for subtree in tree. subtrees():
        print(subtree)
    tree.draw()

sentence = tdf["sentences"].iloc[1]
grammar = "NP: {<DT> ?< JJ> *< NN>}"
chunking(sentence, grammar)
```

```python
# NAMED ENTITY RECOGNITION - used to extract information from unstructured text.
#It classifies entities present in a text into categories like person, object, organization, etc


def named_entity_recognition(text):
    word_tokens = word_tokenize(text)
    word_pos = pos_tag(word_tokens)
    print(ne_chunk(word_pos))

named_entity_recognition(tdf["sentences"].iloc[1])
```
